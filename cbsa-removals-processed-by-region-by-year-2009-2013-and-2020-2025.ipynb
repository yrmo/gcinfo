```python
import gcinfo
import pandas as pd
import matplotlib.pyplot as plt

gcinfo.search_titles("removals")
```

```python
gcinfo.list_resources(gcinfo.search_datasets(query='title:"Removals Processed by Region by Year"')["results"][0])
```

```python
df = gcinfo.download_csv("https://www.cbsa-asfc.gc.ca/data/remove-renvoi-eng.csv")
df.head(20)
```

```python
df["Removals Executed"].unique()
```

```python
regions = [int(x) for x in df.iloc[1, :].tolist()[1:]]
regions
```

```python
total_removals_executed = [int(x.replace(",", "")) for x in df.iloc[8, :].tolist()[1:]]
total_removals_executed
```

```python
df_regions_total_removals_executed = pd.DataFrame({'total': total_removals_executed, 'regions': regions})
df_regions_total_removals_executed
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.bar(regions, total_removals_executed)
 

ax.set_title("Removals Processed by Region")
ax.set_xlabel("Region")
ax.set_ylabel("Total Removals Executed")

plt.xticks(rotation=45, ha="right")
plt.subplots_adjust(bottom=0.3)

plt.figtext(
    0.5, -0.03,
    "Source: Canada Border Services Agency. Removals Processed by Region by Year",
    ha="center",
    fontsize=8,
    style="italic"
)

plt.tight_layout()
plt.show()
```

```python
# Source: Canada Border Services Agency (CBSA) official removals statistics page
# https://www.cbsa-asfc.gc.ca/agency-agence/reports-rapports/security-securite/removals-renvois-eng.html
# Data extracted from the "Removals by region" annual table (calendar years).
# Note: These are enforced removals.
import pandas as pd

# Source: Canada Border Services Agency (CBSA) official removals statistics
# https://www.cbsa-asfc.gc.ca/agency-agence/reports-rapports/security-securite/removals-renvois-eng.html
# "Removals by region" table (calendar years). Open data portal version only goes to 2013.

data = [
    {"Region": "Atlantic",             2020: 57,    2021: 70,   2022: 62,   2023: 104,  2024:  159,   2025: 188},
    {"Region": "Quebec",               2020: 5924,  2021: 1593, 2022: 2496, 2023: 6036, 2024:  6970,  2025: 10628},
    {"Region": "Northern Ontario",     2020: 284,   2021: 234,  2022: 295,  2023: 410,  2024:  491,   2025: 546},
    {"Region": "Greater Toronto Area", 2020: 4500,  2021: 3658, 2022: 2714, 2023: 4306, 2024:  5411,  2025: 7162},
    {"Region": "Southern Ontario",     2020: 204,   2021: 304,  2022: 697,  2023: 1347, 2024:  1072,  2025: 984},
    {"Region": "Prairie",              2020: 966,   2021: 833,  2022: 590,  2023: 772,  2024:  1116,  2025: 1545},
    {"Region": "Pacific",              2020: 923,   2021: 831,  2022: 1481, 2023: 2256, 2024:  2178,  2025: 2107},
    {"Region": "Grand totals",         2020: 12858, 2021: 7523, 2022: 8335, 2023: 15231, 2024: 17397, 2025: 23160}
]

df_report = pd.DataFrame(data)
df_report = df_report.set_index("Region")
df_report
```

```python
df_report.iloc[0:7].sum(axis=0)
```

```python
removals_2020_2025 = list(int(x) for x in df_report.iloc[[7]].to_numpy()[0])
```

```python
regions + list(range(2020, 2026))
```

```python
total_removals_executed + removals_2020_2025
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

fig, ax = plt.subplots()

x = regions + list(range(2020, 2026))
y = total_removals_executed + removals_2020_2025

ax.bar(x, y)

ax.set_title("CBSA Total Enforced Removals by Year")
ax.set_xlabel("Year")
ax.set_ylabel("Total Enforced Removals")

# Force clean integer year labels
ax.xaxis.set_major_locator(MaxNLocator(integer=True))
ax.set_xticks(x)                       # optional: show a tick at every bar
ax.set_xticklabels([str(int(val)) for val in x], rotation=45, ha="right")

plt.subplots_adjust(bottom=0.3)

plt.figtext(
    0.5, -0.05,
    "Sources: CBSA Open Data – Removals Processed by Region by Year\n"
    "CBSA Immigration Removal Statistics, Table 1.1 (2020–2025)",
    ha="center",
    fontsize=8,
    style="italic"
)

plt.tight_layout()
plt.show()
```
