```python
import pandas as pd
import matplotlib.pyplot as plt

def get_total_series(url):
    df = pd.read_excel(url, header=None)
    
    # years are in row 2 (0-indexed), starting at column 4
    years = df.iloc[2, 4:].astype(int)
    
    # the Canada-wide "Total" row
    total_row = df[df.iloc[:, 0].astype(str).str.strip() == "Total"].iloc[0, 4:]
    
    # clean the numbers (commas → int, "--" → NaN)
    totals = (
        total_row
        .astype(str)
        .str.replace(",", "", regex=False)
        .replace("--", pd.NA)
        .astype("Int64")
    )
    
    return pd.Series(totals.values, index=years, name="Total")

imp = get_total_series(
    "https://www.ircc.canada.ca/opendata-donneesouvertes/data/EN_ODP_annual-TR-work-IMP_PT_program_year_end.xlsx"
)
tfw = get_total_series(
    "https://www.ircc.canada.ca/opendata-donneesouvertes/data/EN_ODP_annual-TR-work-TFW_PT_program_year_end.xlsx"
)
```

```python
print("IMP Total:")
print(imp)
print("\nTFW Total:")
print(tfw)
```

```python
fig, ax = plt.subplots()
ax.plot(imp.index, imp, label="IMP", linestyle="-")
ax.plot(tfw.index, tfw, label="TFW", linestyle="-")
ax.set_xlabel("Year")
ax.set_ylabel("Permit Holders (Dec 31)")
ax.set_title("Canada TFW and IMP work Permit Holders")
ax.legend(
    title=None,
    loc="upper center",
    bbox_to_anchor=(0.5, -0.22),
    ncol=1,
    frameon=True
)
plt.figtext(
    0.5, -0.1,
    "Source: Immigration, Refugees and Citizenship Canada. Temporary Residents:\nTemporary Foreign Worker Program (TFWP) and International Mobility Program (IMP)\nWork Permit Holders – Monthly IRCC Updates",
    ha="center",
    fontsize=8,
    style="italic"
)
plt.subplots_adjust(bottom=0.3)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
