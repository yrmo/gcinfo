```python
import matplotlib.pyplot as plt
import numpy as np

# Extract quarter totals
headers = df_xlsx.iloc[3]
data_row = df_xlsx.iloc[-7]

mask = headers.str.contains("Total", na=False)
q_values_raw = data_row[mask].tolist()
q_values = [int(str(v).replace(",", "")) for v in q_values_raw]

# Build labels starting at 2015 Q1
start_year = 2015
q_labels = []
for i in range(len(q_values)):
    year = start_year + (i // 4)
    quarter = (i % 4) + 1
    q_labels.append(f"{year}")

x = np.arange(len(q_values))

fig, ax = plt.subplots()

ax.plot(x, q_values, linestyle="-")

ax.set_title("Canada Work permit holders for Humanitarian and Compassionate")
ax.set_xlabel("Quarter")
ax.set_ylabel("Number of Work Permit Holders")

# Show every 4th label to keep the axis readable
tick_idx = x[::4]
ax.set_xticks(tick_idx)
ax.set_xticklabels([q_labels[i] for i in tick_idx])

plt.xticks(rotation=45, ha="right")
plt.subplots_adjust(bottom=0.3)

plt.figtext(
    0.5, -0.05,
    "Source: Immigration, Refugees and Citizenship Canada.\n"
    "Temporary Residents: Temporary Foreign Worker Program (TFWP) and International Mobility Program (IMP)\n"
    "Work Permit Holders – Monthly IRCC Updates",
    ha="center",
    fontsize=8,
    style="italic"
)

plt.tight_layout()
plt.show()
```
