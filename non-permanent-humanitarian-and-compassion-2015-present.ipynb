```python
import matplotlib.pyplot as plt
import numpy as np

# Quarter totals. A quarter still missing a month is left out.
headers = df_xlsx.iloc[3]
data_row = df_xlsx.iloc[-7]
labels = headers.tolist()
quarter_months = {
    "Q1 Total": ["Jan", "Feb", "Mar"],
    "Q2 Total": ["Apr", "May", "Jun"],
    "Q3 Total": ["Jul", "Aug", "Sep"],
    "Q4 Total": ["Oct", "Nov", "Dec"],
}

q_values = []
for i, label in enumerate(labels):
    if label not in quarter_months:
        continue
    window = [str(v) for v in labels[max(0, i - 8):i] if isinstance(v, str)]
    if not all(month in window for month in quarter_months[label]):
        continue
    q_values.append(int(str(data_row.iloc[i]).replace(",", "")))

# Build labels starting at 2015 Q1
start_year = 2015
q_labels = []
for i in range(len(q_values)):
    year = start_year + (i // 4)
    quarter = (i % 4) + 1
    q_labels.append(f"{year}")

x = np.arange(len(q_values))

fig, ax = plt.subplots()

ax.plot(x, q_values, linestyle="-")

ax.set_title("Canada Work permit holders for Humanitarian and Compassionate")
ax.set_xlabel("Quarter")
ax.set_ylabel("Number of Work Permit Holders")

# Show every 4th label to keep the axis readable
tick_idx = x[::4]
ax.set_xticks(tick_idx)
ax.set_xticklabels([q_labels[i] for i in tick_idx])

plt.xticks(rotation=45, ha="right")
plt.subplots_adjust(bottom=0.3)

plt.figtext(
    0.5, -0.05,
    "Source: Immigration, Refugees and Citizenship Canada.\n"
    "Temporary Residents: Temporary Foreign Worker Program (TFWP) and International Mobility Program (IMP)\n"
    "Work Permit Holders – Monthly IRCC Updates",
    ha="center",
    fontsize=8,
    style="italic"
)

plt.tight_layout()
plt.show()
```
