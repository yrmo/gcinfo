```python
import gcinfo
import pandas as pd
import matplotlib.pyplot as plt

gcinfo.search_titles("non permanent")
```

```python
gcinfo.list_resources(gcinfo.search_datasets(query='title:"Repatriation of Canadian citizens and permanent residents due to corona virus on assisted departures and facilitated flights"')["results"][0])
```

```python
df = gcinfo.download_csv("https://open.canada.ca" + "/data/dataset/e185cdc7-9b8b-4d0d-b26a-6aacf68e0509/resource/51339f12-ae90-4c94-8253-34a85ebb026c/download/repartration-dataset-historical-to-present-data.csv")
df
```

```python
df_english_total_number_returned = df_english.iloc[:, 1:].sum(axis=1)
df_english_total_number_returned
```

```python
df_english_total_number_returned_combined = pd.DataFrame({
    "Country": df["GC Country Short Name English |GC Nom abrégé pays anglais"],
    "Total Returned": df_english_total_number_returned
})
df_english_total_number_returned_combined.sort_values(by="Total Returned", ascending=False)
```

```python
df_english_total_number_returned_combined["Total Returned"].sum()
```

```python
fig, ax = plt.subplots()
bars = ax.bar(
    df_english_total_number_returned_combined["Country"],
    df_english_total_number_returned_combined["Total Returned"]
)

# Label only the top 4 countries (numbers only on bars + names on x-axis)
top4_positions = df_english_total_number_returned_combined["Total Returned"].argsort()[::-1][:4]

# Give a bit more headroom so labels don't hit the title
max_val = df_english_total_number_returned_combined["Total Returned"].max()
ax.set_ylim(0, max_val * 1.18)

# Annotate the bars (number only)
for pos in top4_positions:
    value = df_english_total_number_returned_combined["Total Returned"].iloc[pos]
    ax.annotate(
        f"{value:,.0f}",
        xy=(pos, value),
        xytext=(0, 4),
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontsize=6
    )

# Only show country names for the top 4 on the x-axis
xtick_labels = [""] * len(df_english_total_number_returned_combined)
for pos in top4_positions:
    xtick_labels[pos] = df_english_total_number_returned_combined["Country"].iloc[pos]

ax.set_xticks(range(len(df_english_total_number_returned_combined)))
ax.set_xticklabels(xtick_labels, rotation=45, ha="right")

ax.set_title("Total Number of Canadians Repatriated by Country due to Corona Virus")
ax.set_xlabel("Country")
ax.set_ylabel("Total Number Returned")
plt.subplots_adjust(bottom=0.32)
plt.figtext(
    0.5, -0.05,
    "Source: Global Affairs Canada. Repatriation of Canadian citizens and permanent residents\n"
    "due to corona virus on assisted departures and facilitated flights. Number of Canadian citizens\n"
    "and permanent residents repatriated by country. Operation concluded July 17, 2020",
    ha="center",
    fontsize=8,
    style="italic"
)
plt.tight_layout()
plt.show()
```
