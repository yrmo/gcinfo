```python
import requests
import zipfile
import io
import pandas as pd

url = "https://www150.statcan.gc.ca/n1/tbl/csv/17100121-eng.zip"

r = requests.get(url, timeout=60)
r.raise_for_status()

with zipfile.ZipFile(io.BytesIO(r.content)) as z:
    print(z.namelist())
    files = [n for n in z.namelist() if n.endswith(".csv")]
    with z.open(files[0]) as f: # first file is data file 
        df = pd.read_csv(f)
    with z.open(files[1]) as f: # second file is metadata
        df_metadata = pd.read_csv(f)
```

```python
df.to_csv(files[0])
df_metadata.to_csv(files[1])
```
