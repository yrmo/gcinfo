```python
import requests
import zipfile
import io
import pandas as pd

url = "https://www150.statcan.gc.ca/n1/tbl/csv/17100121-eng.zip"

r = requests.get(url, timeout=60)
r.raise_for_status()

with zipfile.ZipFile(io.BytesIO(r.content)) as z:
    print(z.namelist())
    csv_name = [n for n in z.namelist() if n.endswith(".csv")][0]
    with z.open(csv_name) as f:
        df = pd.read_csv(f)

print(df.shape)
print(df.columns.tolist())
print(df.head())
```
