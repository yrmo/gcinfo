```python
import gcinfo

gcinfo.list_resources(gcinfo.search_datasets(query='title:"Temporary Residents: Temporary Foreign Worker Program (TFWP) and International Mobility Program (IMP) Work Permit Holders – Monthly IRCC Updates"')["results"][0])
```

```python
df_xlsx = gcinfo.download_xlsx("https://www.ircc.canada.ca/opendata-donneesouvertes/data/EN_ODP-TR-Work-HC_citizenship_sign.xlsx")
df_xlsx
```

```python
import matplotlib.pyplot as plt

# Extract countries and last column (assumed to be the relevant total per country)
countries = df_xlsx.iloc[4:-7, 0].values
raw_totals = df_xlsx.iloc[4:-7, -1].values

# Clean country names: keep only the part before the first comma
clean_countries = []
for c in countries:
    name = str(c).split(",")[0].strip()
    clean_countries.append(name)

# Clean numeric strings: remove commas, treat '--' / blanks as 0
totals = []
for x in raw_totals:
    s = str(x).replace(",", "").strip()
    if s in ("--", "", "nan", "None", "NaN"):
        totals.append(0)
    else:
        totals.append(int(float(s)))

# Sort descending and keep top 20 for readability
sorted_pairs = sorted(zip(clean_countries, totals), key=lambda x: x[1], reverse=True)[:20]
top_countries, top_totals = zip(*sorted_pairs)

fig, ax = plt.subplots()
ax.bar(top_countries, top_totals)

ax.set_title("Work Permit Holders for Humanitarian and Compassionate Purposes by Country of Citizenship")
ax.set_xlabel("Country of Citizenship")
ax.set_ylabel("Number of Work Permit Holders")

plt.xticks(rotation=45, ha="right")
plt.subplots_adjust(bottom=0.3)

plt.figtext(
    0.5, -0.05,
    "Source: Immigration, Refugees and Citizenship Canada.\n"
    "Temporary Residents: Temporary Foreign Worker Program (TFWP)\nand International Mobility Program (IMP) Work Permit Holders – Monthly IRCC Updates",
    ha="center",
    fontsize=8,
    style="italic"
)

plt.tight_layout()
plt.show()
```
