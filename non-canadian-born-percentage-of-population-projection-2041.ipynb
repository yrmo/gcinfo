```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.plot(df["Year"], df["Immigrants"], linestyle="-", color="black", label="Non-Canadian Born Population")
ax.set_xlabel("Year")
ax.set_ylabel("Non-Canadian Born Population")

ax2 = ax.twinx()
ax2.plot(
    df["Year"],
    df["Percentage"],
    linestyle="--",
    # marker=".",
    color="black",
    label="Non-Canadian Born Percentage of Population"
)
ax2.set_ylabel("Non-Canadian Born Percentage of Population")

# Combine legends from both axes
lines1, labels1 = ax.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax.legend(
    lines1 + lines2,
    labels1 + labels2,
    title=None,
    loc="upper left",
    frameon=True
)

ax.set_title("Non-Canadian Born Percentage of Population Projection 2041")

# Show every year, rotate labels 45 degrees, and make them slightly smaller
ax.set_xticks(df["Year"])
ax.tick_params(axis="x", rotation=45, labelsize=8)
plt.setp(ax.get_xticklabels(), ha="right")

plt.subplots_adjust(bottom=0.25)
plt.figtext(
    0.5, -0.075,
    "Source: Census of Population, 1871 to 2006, 2016 and 2021 (3901);\n"
    "National Household Survey, 2011 (5178); Population projections on immigration\n"
    "and diversity for Canada and its regions, 2016 to 2041 (5126)",
    ha="center",
    fontsize=8,
    style="italic"
)
plt.tight_layout()
# plt.savefig(fname="2041.png", bbox_inches='tight')
plt.show()
```
