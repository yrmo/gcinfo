```python
import gcinfo
import pandas as pd
import matplotlib.pyplot as plt

gcinfo.search_titles("removals")
```

```python
gcinfo.list_resources(gcinfo.search_datasets(query='title:"Removals Processed by Region by Year"')["results"][0])
```

```python
df = gcinfo.download_csv("https://www.cbsa-asfc.gc.ca/data/remove-renvoi-eng.csv")
df.head(20)
```

```python
df["Removals Executed"].unique()
```

```python
regions = [int(x) for x in df.iloc[1, :].tolist()[1:]]
regions
```

```python
total_removals_executed = [int(x.replace(",", "")) for x in df.iloc[8, :].tolist()[1:]]
total_removals_executed
```

```python
df_regions_total_removals_executed = pd.DataFrame({'total': total_removals_executed, 'regions': regions})
df_regions_total_removals_executed
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.bar(regions, total_removals_executed)
 

ax.set_title("Removals Processed by Region")
ax.set_xlabel("Region")
ax.set_ylabel("Total Removals Executed")

plt.xticks(rotation=45, ha="right")
plt.subplots_adjust(bottom=0.3)

plt.figtext(
    0.5, -0.03,
    "Source: Canada Border Services Agency. Removals Processed by Region by Year",
    ha="center",
    fontsize=8,
    style="italic"
)

plt.tight_layout()
plt.show()
```
